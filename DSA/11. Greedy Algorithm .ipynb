````python
from typing import List
class Solution:
    def findContentChildren(self, g: List[int], s: List[int]) -> int:
        g.sort()
        s.sort()
        i=j=0
        counter=0
        while i<len(g) and j<len(s):
            if g[i]<=s[j]:
                counter=counter+1
                i=i+1
                j=j+1
            else:
                j=j+1
        return counter
````

````python
class Solution:
    def findMin(self, n):
        output = 0
        if n >= 10:
            output += (n) // 10
            n = n % 10
        if n >= 5:
            output += (n) // 5
            n = n % 5
        if n >= 2:
            output += (n) // 2
            n = n % 2
        if n >= 1:
            output += n   # n itself is number of 1-rupee coins
        return output
````

````python
class Solution:
    def lemonadeChange(self, bills):
        five = 0
        ten = 0

        for bill in bills:
            if bill == 5:
                five += 1

            elif bill == 10:
                if five == 0:
                    return False
                five -= 1
                ten += 1

            else:  # bill == 20
                if ten > 0 and five > 0:
                    ten -= 1
                    five -= 1
                elif five >= 3:
                    five -= 3
                else:
                    return False

        return True
````

# 🍋 Lemonade Change – State vs Sum (Core Insight)

## 🔑 What you were doing (money-based thinking)

You were thinking like this:

* 5 → +5
* 5 → +5
* 5 → +5

**Total = 15**

Then a **20** comes:

* I take 20
* I give 15 as change
* So I still *“have”* 5

So your mental state became:

```
Money = 5
```

---

## ❌ This is where the illusion happens

That **₹5** is **imaginary** — it does **not** correspond to any real bill left in your hand.

You are doing **math**, not **cashier logic**.

---

## 🔥 The reality (bill-based thinking)

### What you ACTUALLY had before the 20 came

```
Bills in hand = [5, 5, 5]
```

### Customer pays 20 → needs 15 change

You give:

```
5 + 5 + 5
```

### After giving change

```
Bills in hand = []
```

👉 **Zero bills remain**

Even though mathematically:

```
20 - 15 = 5
```

That **₹5 never exists physically**.

---

## 💡 THE CORE LESSON (very important)

Cashier problems are about **STATE**, not **SUM**.

**State =**

* how many ₹5 notes
* how many ₹10 notes

**Not:**

* total money

---

## 🧠 Your corrected mental model

❌ Wrong:

```
change = total money
```

✅ Correct:

```
five = number of ₹5 notes
ten  = number of ₹10 notes
```

That’s why the correct solution tracks **notes**, not money.

````python
class Solution:
    def checkValidString(self, s: str) -> bool:
        
        # Forward pass
        star = 0
        left = 0
        for char in s:
            if char == "*":
                star += 1
            elif char == ")":
                if left > 0:
                    left -= 1
                elif star > 0:
                    star -= 1
                else:
                    return False
            else:
                left += 1
        if left > star:
            return False

        # Reverse pass
        star = 0
        right = 0
        for char in reversed(s):
            if char == "*":
                star += 1
            elif char == "(":
                if right > 0:
                    right -= 1
                elif star > 0:
                    star -= 1
                else:
                    return False
            else:
                right += 1
        
        return False if right > star else True
````

# Why must 0 lie inside the range `[min, max]` at the END?

### 🎯 Core Idea

The string is valid **only if the final balance of open brackets becomes 0**.

`min = minimum possible open brackets`
`max = maximum possible open brackets`

For a valid string:

**0 must lie inside the range:**

```
min <= 0 <= max
```

If 0 is *not* inside the range, the string can never become completely balanced.

---

## ❌ Example: Range = [1, 4]

This means:

* **Minimum open brackets = 1** → Even if all `*` are treated as `)` the best you can do still leaves 1 unclosed `(`.
* **Maximum open brackets = 4** → If all `*` are treated as `(` you reach 4.

So the possible ending balances are:

```
1, 2, 3, 4
```

None of these include **0**, which is **required** for a valid parenthesis sequence.

Hence, **the string is invalid**.

---

## 🔥 Why is 0 so important?

* A valid parenthesis string MUST end with **0 open brackets**.
* If final balance > 0 → some `(` are left unmatched → invalid.

The `min/max` strategy ensures:

```
min <= actual_balance <= max
```

But if at the END:

```
min > 0
```

Then:

* Even after the most aggressive closing (`*` → `)`), we **cannot** reach balance = 0.
* So no possible interpretation can make the string valid.

---

### ✔️ Final Summary

A valid string must end balanced.
Balance = 0.
Therefore:

```
0 must lie in [min, max]
```

If not → no valid interpretation exists → ❌ invalid string.


````python
class Solution:
    def checkValidString(self, s: str) -> bool:
        min_open = 0
        max_open = 0
        for char in s:
            if char == '(':
                min_open += 1
                max_open += 1
            elif char == ')':
                min_open -= 1
                max_open -= 1
             # If current character is '*', it can be '(', ')' or ''
            else:
                 # If '*' acts as ')'
                min_open -= 1    
                # If '*' acts as '('
                max_open += 1     
            if max_open < 0:
                return False
            if min_open < 0:
                min_open = 0
        print(min_open)
        return min_open == 0

s = '(**('
sol = Solution()
ans=sol.checkValidString(s)
ans
````

````python
class Solution:
    def maximumMeetings(self,start,end):
        meeting_data=[]
        result=[]
        last_meet_ends=-1
        for i in range(len(start)):
          meeting_data.append((end[i],start[i],i))
        meeting_data.sort()
        for e, s, idx in meeting_data:
            if s > last_meet_ends:  
              result.append(idx) 
              last_meet_ends = e  
        return len(result)
````

#### Brute force approach (recursion)

````python
class Solution:
    def canJump(self, nums: List[int]) -> bool:
      n=len(nums)
      def canReach(i):
        if n-1<=i:
          return True
        if nums[i]==0:
           return False

        for jump in range(1,nums[i]+1):
          if canReach(i+jump):
             return True

        return False
      return canReach(0) 
         
````

#### Optimal approach (Greedy)

````python
class Solution:
    def canJump(self, nums: List[int]) -> bool:
        maxIndex=nums[i]
        for i in range(1,len(nums)):
            if i<=maxIndex:
                maxIndex=max(i+nums[i],maxIndex)
            else:
                return False
        return True
            
````

````python
from typing import List
class Solution:
    def jump(self, nums: List[int]) -> int:
        return self.minJump(nums, 0)
    def minJump(self,nums,position):
        if position >= len(nums) - 1:
            return 0
        min_step = float('inf')
        for jump in range(1,nums[position]+1):
            ans=self.minJump(nums,position+jump)
            if ans != float('inf'):
                min_step = min(min_step, 1 + ans)
        return min_step

sol=Solution()
ans=sol.jump([2,3,1,1,4])
print(ans)
````

````python
class Solution:
    def jump(self, nums: List[int]) -> bool:
        jump=0
        l,r=0,0
        while (r<(len(nums)-1)):
            farthest=0
            for i in range(l,r+1):
                farthest=max(farthest,i+nums[i])
            l=r+1
            r=farthest
            jump=jump+1
        return jump
            
````

`2.` `Optimal`

````python
class Solution:
    def jobSequencing(self, deadline, profit):
        jobs = sorted(zip(profit, deadline), reverse=True)
        max_deadline = max(deadline)
        parent = list(range(max_deadline + 1))

        def find(x):
            if parent[x] == x:
                return x
            parent[x] = find(parent[x])
            return parent[x]

        count = 0
        total_profit = 0
        for p, d in jobs:
            slot = find(d)
            if slot > 0:
                parent[slot] = find(slot-1)
                count += 1
                total_profit += p
        return [count, total_profit]
````

    2.6 Candy
   </a>
</p>


`1.` `In Two pass (Brute force)` `O(N)`

````python
class Solution:
    def candy(self, ratings):
        n = len(ratings)
        candies = [1] * n
        # left → right
        for i in range(1,n):
            if ratings[i] > ratings[i-1]:
                candies[i] = candies[i-1] + 1
        # right → left
        for i in range(n-2,-1,-1):
            if ratings[i] > ratings[i+1]:
                candies[i] = max(candies[i], candies[i+1] + 1)
        return sum(candies)
````

`2.` `In One pass -> using slope / mountain approach (Optimal)` `O(1)`

````python
class Solution:

    def candy(self, ratings):
        n = len(ratings)
        candies = 1
        i = 1

        while i < n:
            # skip equal ratings
            if ratings[i] == ratings[i-1]:
                candies=candies+1
                i += 1
                continue
            peak = 1
            # increasing slope
            while i < n and ratings[i] > ratings[i-1]:
                peak += 1
                candies += peak
                i += 1
            down = 1
            # decreasing slope
            while i < n and ratings[i] < ratings[i-1]:
                candies += down
                i += 1
                down=down+1
                if down>peak:
                    candies=candies+(down-peak)

            # remove double counted peak
            
        return candies

ratings = [1,2,2]
sol = Solution()
print(sol.candy(ratings))
````

````python
class Solution:
    def insert(self, intervals, newInterval):
        result=[]
        for i in range(len(intervals)):
            # for case 1: [[1,2],[3,5]] and newint=[4,8] then newInt should come after [1,2] and later it will go to else case for overlap
            if intervals[i][1]<newInterval[0]:
                result.append(intervals[i])

            # for case 2: [[1,2],[9,5]] and newint=[4,8] then newInt should come after [1,2] and before [9,5] so we insert it and make newInt = [9,5] and later it will append to result at last
            elif intervals[i][0]>newInterval[1]:
                result.append(newInterval)
                newInterval=intervals[i]

            # Overlaping case inculde case 1 when we have to merge overlapping intervals. 
            else:
                newInterval[0] = min(newInterval[0], intervals[i][0])
                newInterval[1] = max(newInterval[1], intervals[i][1])

        # Inculdes case 2 when we have to add newInt in middle
        result.append(newInterval)
        return result
````

````python
from typing import List

class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        if len(intervals)==1:
            return intervals
        intervals.sort()
        result=[]
        over_lapping=intervals[0]
        for i in range(1,len(intervals)):
            if intervals[i][0]<=over_lapping[1]:
                if intervals[i][1]>over_lapping[1]:
                    over_lapping[1]=intervals[i][1]
                # Or
                # over_lapping[1] = max(over_lapping[1], intervals[i][1])
            else:
                result.append(over_lapping)
                over_lapping=intervals[i]
        result.append(over_lapping)
        return result
````

Non-overlapping Intervals

````python
from typing import List

class Solution:
    def eraseOverlapIntervals(self, intervals: List[List[int]]) -> int:
        intervals.sort(key=lambda x: x[1])  # sort by end time
        result = 0
        prev_end = intervals[0][1]
        for i in range(1, len(intervals)):
            if intervals[i][0] < prev_end: 
                result += 1
            else:
                prev_end = intervals[i][1]
        return result
````

````python
from typing import List

class Solution:
    def eraseOverlapIntervals(self, intervals: List[List[int]]) -> int:
        result = 0
        data = []
        for i in range(len(intervals)):
            start = intervals[i][0]
            end = intervals[i][1]
            data.append((end, start))
        data.sort()
        prev_end = data[0][0]
        for i in range(1, len(data)):
            start = data[i][1]
            end = data[i][0]
            if start < prev_end:
                result += 1
            else:
                prev_end = end
        return result
````
